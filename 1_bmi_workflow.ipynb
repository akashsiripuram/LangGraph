```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BMIState(TypedDict):
    weight_kg:float
    height_m:float
    bmi:float
    category:str
```

```python
#node bmi calculator
def bmi_calculator(state:BMIState)->BMIState:
    
    weight=state['weight_kg']
    height=state['height_m']
    
    bmi=weight/(height**2)
    
    state['bmi']=round(bmi,2)
    return state
```

```python
def category_bmi(state: BMIState) -> BMIState:
    bmi = state['bmi']

    if bmi < 18:
        state['category'] = 'Underweight'
    elif bmi < 25:
        state['category'] = 'Normal weight'
    elif bmi < 30:
        state['category'] = 'Overweight'
    else:
        state['category'] = 'Obese'

    return state
```

```python
# Define graph
graph=StateGraph(BMIState)

#add nodes to the graph
graph.add_node('bmi_calculator',bmi_calculator)
graph.add_node('category_bmi',category_bmi)

#add edges between the nodes
graph.add_edge(START,"bmi_calculator")
graph.add_edge("bmi_calculator","category_bmi")
graph.add_edge("category_bmi",END)

#compile the graph
workflow=graph.compile()
```

```python
initial_state={"weight_kg":50,"height_m":1.67}

final_state=workflow.invoke(initial_state)

print(final_state)
```
