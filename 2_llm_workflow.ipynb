```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class State(TypedDict):
    question:str
    answer:str
```

```python
from dotenv import load_dotenv
load_dotenv()
#node llm_answer
def llm_answer(state:State)->State:
    question = state['question']

    llm=ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')
    
    state['answer']=llm.invoke(question).content[0]['text']
    return state
```

```python
#define graph
graph=StateGraph(State)

#add nodes
graph.add_node("llm_answer",llm_answer)

#add edges
graph.add_edge(START,"llm_answer")
graph.add_edge("llm_answer",END)

#compile
workflow=graph.compile()
```

```python
initial_state={"question":"What is Agentic AI"}
final_state=workflow.invoke(initial_state)
print(final_state['answer'])
```
